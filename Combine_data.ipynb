## **Import Library**

```python
import pandas as pd
```

## **Load Data**

---

```python
# Load legal service standard data from CSV into a DataFrame
df_hukum = pd.read_csv('Data/standar_layanan_hukum.csv')

# Display the first few rows for a quick sanity check
df_hukum.head()
```

```python
# Load civil service standard data from CSV into a DataFrame
df_perdata = pd.read_csv('Data/standar_layanan_perdata.csv')

# Display the first few rows for a quick sanity check
df_perdata.head()
```

```python
# Load criminal case service standard data into a DataFrame
df_pidana = pd.read_csv('Data/standar_layanan_pidana.csv')

# Display the first few rows for a quick sanity check
df_pidana.head()
```

```python
# Load general service standard data into a DataFrame
df_umum = pd.read_csv('Data/standar_layanan_umum.csv')

# Display the first few rows for a quick sanity check
df_umum.head()
```

```python
# Load general information FAQ data from CSV into a DataFrame
df_informasi = pd.read_csv('Data/FAQ_informasi_umum.csv')

# Show the first few rows for a quick inspection
df_informasi.head()
```

```python
# Combine all service and FAQ DataFrames into a single unified DataFrame
combined_df = pd.concat([df_hukum, df_perdata, df_pidana, df_umum, df_informasi], ignore_index=True)

# Save the combined dataset to CSV without including the index column
combined_df.to_csv('Data/standar_layanan_combined.csv', index=False)
```
